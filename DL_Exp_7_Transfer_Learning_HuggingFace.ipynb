```python
# DL Experiment 7: Transfer Learning using Hugging Face

# Install transformers
!pip install transformers -q

# Import required library
from transformers import pipeline

# Load pre-trained sentiment analysis model
classifier = pipeline("sentiment-analysis")

# Sample input sentences
sentences = [
    "I absolutely love this new software! It's incredibly intuitive and powerful.",
    "The customer service was terrible; I waited for hours and got no resolution.",
    "This movie was just okay, nothing special, but not bad either.",
    "I am so excited for the upcoming concert!",
    "What a frustrating experience, I will never buy from them again."
]

# Perform predictions
print("\n--- Sentiment Analysis Predictions ---\n")

for sentence in sentences:
    result = classifier(sentence)[0]
    label = result['label']
    score = result['score']

    print(f"Input: \"{sentence}\"")
    print(f"Output: {label} (Confidence: {score:.4f})")
    print("-" * 40)

# Conclusion
print("\nConclusion:")
print("The pre-trained Hugging Face model successfully performed sentiment classification using transfer learning.")
```
